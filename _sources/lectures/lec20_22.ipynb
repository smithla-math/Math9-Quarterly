### Constant List
__Mini Goal:__ Write a function that takes as input a natural number $n$, and as output returns a length $n$ list `[3,3,3,3...]`

In this section, we'll learn the basic syntax for writing functions in python.

```python
def f(n):
    mylist = []
    for _ in range(n):
        mylist.append(3)
```

Notice in the for-loop above, the variable `i` is never used in the loop. In this case, we often use `_` as the variable instead, to indicate it won't show up.

Let's try running our function.

```python
print(f(9))
```

Notice there is no error, but there's also no output! Here's how we can specify the output. The code below still has a small issue.

```python
def f(n):
    mylist = []
    for _ in range(n):
        mylist.append(3)
        return mylist
```

```python
f(10)
```

```python
f(5)
```

What's going on here? Notice that because `return` is in the for-loop, we exit the function right after we append one 3 to `mylist`. To fix this, we have to move `return` outside of the for-loop.

```python
def constant(n):
    mylist = []
    for _ in range(n):   #underscore is useful when you don't use the variable in the loop
        mylist.append(3)
    return mylist


constant(20)
```

Finally, our function should work as expected.

```python
constant(0)
```

```python
constant(7)
```

Notice I ran `constant(7)` above. Within the function `n` was set to 7. But `n` does not persist outside of the function.

```python
n
```

What if we define `n` outside of the function?

```python
n = 100
constant(4) #this line does not affect n, even though we used the label n when we wrote our function.
```

```python
n #n is still 100
```

The above shows us that input variables are locally defined.

### Alternating List - Version 1

Now lets write our alternating function. Our strategy is to first create a length $n$ list of 3s, and then change the odd indices to be 7s. We already have a `constant(n)` function, so let's use it inside of our code for `alt1(n)`.

```python
def alt1(n):
    mylist = constant(n) #reuse the constant function from above.
    for i in range(1,n,2):
        mylist[i] = 7
    return mylist

w = alt1(8)
print(w)
```

Note that calling `constant(n)` inside of `alt1(n)`, has many advantages. For one, it's less work than copying and pasting. Even further, if we realized we need to change `constant(n)`, we only need to do it in one place. If we copy-paste to multiple locations, we have to update the code in all of these locations.

```python
print(w[1::2])
w[1::2] = 8  #notation like this doesn't work for Python lists. Need Python arrays to broadcast (later in this page)
print(w)
```

```python
w[1::2] = [8,8,8,8] #with a list, we need the list to be the same size as the slice. There's a better way to do this, we'll see today.
w
```

### Alternating List - Version 2

Let's see another way of writing a function that does the same thing as `alt1(n)`. The strategy this time is to check whether an index is even or odd.

__Recall:__ We can use `m%n` to calculate m mod n. This returns the remainder of `m` divded by `n`.

```python
print(5%2)
print(7%2)
print(4%2)
```

__Observe:__ even numbers have a remainder of 0 when we divide by 2, and odd numbers have a remainder of 1 when we divide by 2.

```python
def alt2(n):
    mylist = []
    for i in range(n):
        if i%2 == 0:     #i mod 2 equals 0 is the same as saying i has 0 remainder when we divide by 2. This means i is even.
            mylist.append(3)
        else:
            mylist.append(7)
    return mylist
```

```python
print(alt2(11))
```

Notice how we didn't need to specify an `elif`. This is because an integer only has two possible remainders when we divide by 2. If the remainder is not 0, it must be 1.

```python
def cycle_three(n):  #gives lists of the form [3, 7, 6, 3, 7, 6, ... ]
    mylist = []
    for i in range(n): #i represents the index of mylist
        if i%3 == 0:  #if i divided by 3 has zero for a remainder. Same as if i were a multiple of 3.
            mylist.append(3)
        elif i%3 == 1:
            mylist.append(7)
        else:
            mylist.append(6)
    return mylist
```

Now, we test `alt2(n)` and `cycle_three(n)` a few times.

```python
alt2(21)
```

```python
alt2(6)
```

```python
cycle_three(4)
```

```python
cycle_three(20)
```

### Timing Alternating List Strategies
__Upshot:__ NumPy will be much faster than base python!

Let's start by copy-pasting our code for `alt3(n)` as `alt_np(n)`. Remember, this is the code that used NumPy.

```python
import numpy as np
```

```python
def alt_np(n):
    A = np.zeros(n,dtype=np.int64) + 3 #generates array of all 3s
    A[1::2] = 7 #reassigns the odd indices to 7
    return A
```

To get a more accurate time comparison, we're going to write another function `alt_list(n)` which uses lists and is meant to mimic how `alt_np(n)` is structured:
* Create a length n list of all 3s.
* Change all of the odd indices to be 7s.

```python
def alt_list(n):
    mylist = []
    for i in range(n):
        mylist.append(3)  #the first loop generates a list of all 3s
    for i in range(1,n,2):
        mylist[i] = 7 #the second loop reassigns the odd indices to 7
    return mylist
```

Let's run a couple of test cases.

```python
alt_np(10)
```

```python
alt_list(9)
```

Recall, we can use the Jupyter magic `%%timeit` to time how long it takes a block of code to run.

```python
%%timeit 
alt_list(10)
```

We will now see a timing method that gives just the time, without all of the extra information given by `%%timeit`.

```python
#comes with python, no installation
import time 
```

```python
time.time()
```

This gives how much time has elapsed since the beginning of Unix time (computer time): January 1st, 1970.

```python
time.time()
```

__Notice:__ it wouldn't make sense for us to increment `n = 2*n` at the end of the while-loop. Why?

Now, let's see how much faster `alt_np(n)`, our NumPy version, can run for the same `n`.

```python
print(n)
start = time.time()
alt_np(n)
end = time.time()
t = end-start
print(t)
```

Appreciate the difference! What took base python (lists) more than 5 seconds, took NumPy about $\frac{1}{3}$ of a second.

Let's now keep our NumPy code, but return a list at the very end. 

```python
def alt_convert(n):
    A = np.zeros(n,dtype=np.int64) + 3
    A[1::2] = 7
    return list(A)   #Convert numpy array to a list
```

```python
print(n)
start = time.time()
alt_convert(n)
end = time.time()
t = end-start
print(t)
```

What made `alt_convert(n)` so slow? It was the fact that we converted `A` to a list.

What if we used a for-loop instead of slicing in our NumPy function?

```python
def alt_loop(n):
    A = np.zeros(n,dtype=np.int64) + 3
    for i in range(1,n,2):
        A[i] = 7
    return A
```

```python
print(n)
start = time.time()
alt_loop(n)
end = time.time()
t = end-start
print(t)
```

__Takeaway:__ For-loops are very slow! Avoid them when you can :)

### Loops vs `np.arange`
Construct the $2 \times 5$ matrix 

$
\begin{pmatrix}
2 & 5 & 8 & 11 & 14 \\
17 & 20 & 23 & 26 & 29
\end{pmatrix}
$

__Method 1:__ We notice that the difference between consecutive elements is 3. We'll use a for-loop to create this array.

```python
arr=np.zeros((2,5))
arrshape = arr.shape #The shape gives the dimensions we need for our loops
print(arrshape[0])
print(arrshape[1])
for row in range(arrshape[0]):
    for col in range(arrshape[1]):
        arr[row,col] = 2 + 15*row + 3*col
    print(arr)
```

This version will adapt to the size of the array you give it.

```python
arr=np.zeros((3,10))
arrshape = arr.shape #The shape gives the dimensions we need for our loops
print(arrshape[0])
print(arrshape[1])
for i in range(arrshape[0]):
    for j in range(arrshape[1]):
        arr[i,j] = 2 + 3*arrshape[1]*i + 3*j
    print(arr)
```
