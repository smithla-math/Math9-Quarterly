In this exercise, we will explore how polynomial regression can be used to fit data.

In particular, the functions `np.polyfit` and `np.poly1d` will be very useful here.

Suppose we collected the following data points:

- `x = [1,2,3,4,5]` 
- `y = [2,4,5,4,5]`

Questions
1. Fit a degree 2 polynomial to the data.
2. Plot the data points.
3. Plot the polynomial regression curve.
4. Based on the plot, does the model fit the data well?

Try modifying the degree of the polynomial to see how the curve changes.


Code: 

```python
import numpy as np
import matplotlib.pyplot as plt

#fixed data example
#x = np.array([1,2,3,4,5])
#y = np.array([2,4,5,4,5])

#randomized data
rng = np.random.default_rng()
x = rng.integers(-10,10,size=5)
y = rng.integers(-10,10,size=5)
print(f" x = {x}, \n y = {y}")

# polynomial regression
degree = 2
coeffs = np.polyfit(x, y, degree) #np.polyfit(x,y,degree) Try degree = 1, 2, 3, 4

poly = np.poly1d(coeffs) #poly will be the name of the function we generated
x_line = np.linspace(-10,10,100)
y_line = poly(x_line)

fig, ax = plt.subplots()
ax.scatter(x, y)
ax.plot(x_line, y_line)

fig.suptitle(f"Polynomial Regression Example, degree = {degree}")
plt.xlabel("x")
plt.ylabel("y")

plt.show()
```
